DATA CLEANING

```python
df = df.drop_duplicates()
```

```python
num_col = df.select_dtypes(include="number").columns
for col in num_col :
    df[col] = df[col].fillna(df[col].mean())
```

```python
cat_cols
```

```python
for col in cat_cols :
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
for col in cat_cols :
    df[col] = df[col].str.strip()
```

```python
for col in num_col:
    sns.boxplot(x=df[col])
    plt.show()
```

HISTOGRAM
